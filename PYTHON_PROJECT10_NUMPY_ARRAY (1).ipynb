1.	Write a Python program to create a NumPy array from a list and print its shape.
o	List = [1, 2, 3, 4, 5, 6]

5.	Write a Python program to convert the data type of a NumPy array from float to int using astype().
o	Array = [5.5, 6.7, 8.9]


```python
import numpy as np

arr = np.array([5.5, 6.7, 8.9])

# Convert float array to integer array
int_arr = arr.astype(int)

print("Original Array:", arr)
print("Converted Integer Array:", int_arr)
print("Datatype after conversion:", int_arr.dtype)
```
