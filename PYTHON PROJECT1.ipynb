10. Write a Python program to swap two numbers entered by the user without using a third variable.


```python
# with easy method
a = int(input("enter a first number="))
b = int(input("enter a second number="))
a,b = b,a
print("a =",a)
print("b =",b)
```

```python
# with arithematic method
a = int(input("enter a first number="))
b = int(input("enter a second number="))

a = a+b
b = a-b
a = a-b
print("a =",a)
print("b =",b)
```

Take two numbers as input and find their greatest common divisor (GCD)

```python
import math
A = int(input("enter a first number="))
B = int(input("enter a second number="))

gcd = math.gcd(A, B)
print(f"gcd of {A} and {B} is {gcd}")
```

Write a Python program to calculate the simple interest using the formula: SI = (P × R × T) / 100, where P is the principal amount, R is the rate of interest, and T is the time in years

```python
P = int(input("enter a principal amount="))
R = int(input("enter a rate of interest="))
T = int(input("enter a time in years="))
SI = (P*R*T)/100
print("The simple interest=", SI)
```

Write a Python program to check whether a number is positive, negative, or zero.


```python
num = float(input("Enter a number: "))

if num > 0:
    print(num, "is Positive")
elif num < 0:
    print(num, "is Negative")
else:
    print("The number is Zero")
```

**19.Write a Python program to calculate the perimeter of a circle using the formula: perimeter = 2 * π * radius.**


```python
import math

radius = float(input("Enter the radius of the circle: "))

perimeter = 2 * math.pi * radius

print("Perimeter of the circle is:", perimeter)
```
